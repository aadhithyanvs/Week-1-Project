```python
numeric_cols = df.select_dtypes(include='number').columns
```

```python
Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_counts = (
    (df[numeric_cols] < lower_bound) |
    (df[numeric_cols] > upper_bound)
).sum()

print(outlier_counts)
```

```python
print(lower_bound)
print(upper_bound)
selected_cols = ['MonthlyIncome', 'TotalWorkingYears',
                 'YearsAtCompany', 'TrainingTimesLastYear']

for col in selected_cols:
    print(col, ":", outlier_counts[col])
```

```python
print(df.loc[
    (df['TotalWorkingYears'] < lower_bound['TotalWorkingYears']) |
    (df['TotalWorkingYears'] > upper_bound['TotalWorkingYears']),
    'TotalWorkingYears'
])
```


### Outlier Handling Decision

Outliers were identified using the IQR method.

After examining the potential outliers, all records were retained because the extreme values may represent valid employee information.

No rows were removed based solely on the IQR results.


```python

clean_df = df.copy()

for col in clean_df.select_dtypes(include='object').columns:
    clean_df[col] = clean_df[col].str.strip()

for col in clean_df.select_dtypes(include='object').columns:
    clean_df[col] = clean_df[col].str.title()

print(clean_df.dtypes)

for col in clean_df.select_dtypes(include='object').columns:
    print(col, ":", clean_df[col].unique())
```

```python

print(clean_df.dtypes)

print(clean_df.select_dtypes(include='object').nunique())

print(clean_df.duplicated().sum())

clean_df = clean_df.drop_duplicates()

print(clean_df.shape)

print(clean_df.isna().sum().sum())
```

```python

missing_values = clean_df.isna().sum()

print(missing_values[missing_values > 0])

clean_df['MonthlyIncome'] = clean_df['MonthlyIncome'].fillna(clean_df['MonthlyIncome'].median())

clean_df['NumCompaniesWorked'] = clean_df['NumCompaniesWorked'].fillna(clean_df['NumCompaniesWorked'].median())

clean_df['TotalWorkingYears'] = clean_df['TotalWorkingYears'].fillna(clean_df['TotalWorkingYears'].median())

clean_df['TrainingTimesLastYear'] = clean_df['TrainingTimesLastYear'].fillna(clean_df['TrainingTimesLastYear'].median())

print(clean_df.isna().sum().sum())
```

```python

print(clean_df.isnull().sum())

print("Total missing values:", clean_df.isnull().sum().sum())

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", clean_df.shape)

summary = pd.DataFrame({
    'Before Cleaning': [
        len(df),
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ],
    'After Cleaning': [
        len(clean_df),
        clean_df.isnull().sum().sum(),
        clean_df.duplicated().sum()
    ]
}, index=['Rows', 'Missing Values', 'Duplicates'])

print(summary)

summary.to_csv("../Reports/cleaning_summary.csv")
```

```python

numeric_cols = clean_df.select_dtypes(include='number').columns

summary_stats = clean_df[numeric_cols].agg([
    'mean',
    'median',
    'min',
    'max',
    'std'
])

print(summary_stats)
```

```python

department_analysis = clean_df.groupby('Department').agg({
    'MonthlyIncome': ['mean', 'median'],
    'Age': 'mean',
    'Attrition': lambda x: (x == 'Yes').sum()
})

print(department_analysis)
```

```python

import numpy as np

correlation_matrix = clean_df.select_dtypes(include='number').corr()

upper_triangle = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

top_3_correlations = upper_triangle.unstack().dropna().sort_values(
    key=abs, ascending=False
).head(3)

print(top_3_correlations)
```

```python

#  Which department has the highest employee attrition rate?

department_attrition = clean_df.groupby('Department').agg(
    Total_Employees=('Attrition', 'count'),
    Employees_Left=('Attrition', lambda x: (x == 'Yes').sum())
)

department_attrition['Attrition_Rate'] = (
    department_attrition['Employees_Left'] /
    department_attrition['Total_Employees'] * 100
)

print(department_attrition.sort_values(
    'Attrition_Rate', ascending=False
))
```

```python

# What is the relationship between monthly income and attrition?

income_attrition = clean_df.groupby('Attrition')['MonthlyIncome'].agg(
    ['count', 'mean', 'median']
)

print(income_attrition)
```

```python

#  Which age group has the highest employee attrition rate?

clean_df['AgeGroup'] = pd.cut(
    clean_df['Age'],
    bins=[0, 25, 35, 45, 55, 100],
    labels=['18-25', '26-35', '36-45', '46-55', '56+']
)

age_attrition = clean_df.groupby('AgeGroup', observed=True).agg(
    Total_Employees=('Attrition', 'count'),
    Employees_Left=('Attrition', lambda x: (x == 'Yes').sum())
)

age_attrition['Attrition_Rate'] = (
    age_attrition['Employees_Left'] /
    age_attrition['Total_Employees'] * 100
)

print(age_attrition.sort_values('Attrition_Rate', ascending=False))
```

```python

# Does overtime affect employee attrition?

overtime_attrition = clean_df.groupby('OverTime').agg(
    Total_Employees=('Attrition', 'count'),
    Employees_Left=('Attrition', lambda x: (x == 'Yes').sum())
)

overtime_attrition['Attrition_Rate'] = (
    overtime_attrition['Employees_Left'] /
    overtime_attrition['Total_Employees'] * 100
)

print(overtime_attrition)
```

```python

# Which job role has the highest average monthly income?

jobrole_income = clean_df.groupby('JobRole').agg(
    Average_Monthly_Income=('MonthlyIncome', 'mean'),
    Median_Monthly_Income=('MonthlyIncome', 'median'),
    Total_Employees=('EmployeeNumber', 'count')
)

print(jobrole_income.sort_values(
    'Average_Monthly_Income', ascending=False
))
```

```python

#  Create a department budget table

department_budget = pd.DataFrame({
    'Department': [
        'Sales',
        'Human Resources',
        'Research & Development'
    ],
    'Annual_Budget': [
        5000000,
        2000000,
        10000000
    ]
})

print(department_budget)
```

```python

#  Merge employee data with department budget

merged_df = pd.merge(
    clean_df,
    department_budget,
    on='Department',
    how='left'
)

print(merged_df.head())

print(merged_df.shape)
```

```python

#  Check the merged dataset

print(merged_df.shape)

print(merged_df['Annual_Budget'].isnull().sum())

print(merged_df.groupby('Department')['Annual_Budget'].first())
```

```python

# Create a pivot table for department-wise attrition

department_pivot = pd.pivot_table(
    clean_df,
    index='Department',
    columns='Attrition',
    values='EmployeeNumber',
    aggfunc='count',
    fill_value=0
)

print(department_pivot)
```

```python

# Import visualization libraries

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
```

```python

# 1. Bar Chart – Employee Count by Department

plt.figure(figsize=(8, 5))

sns.countplot(
    data=clean_df,
    x='Department',
    color='#4472C4'
)

plt.title('Employee Count by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=15)

plt.savefig(
    '../Reports/figures/employee_count_by_department.png',
    bbox_inches='tight'
)

plt.show()

# Insight
highest_department = clean_df['Department'].value_counts().idxmax()
highest_count = clean_df['Department'].value_counts().max()

print(f"Insight: {highest_department} has the highest employee count with {highest_count} employees.")
```

```python

# 2. Histogram – Distribution of Employee Ages

plt.figure(figsize=(8, 5))

sns.histplot(
    data=clean_df,
    x='Age',
    bins=15,
    color='#4472C4',
    edgecolor='black'
)

plt.title('Distribution of Employee Ages')
plt.xlabel('Age (Years)')
plt.ylabel('Number of Employees')

plt.savefig(
    '../Reports/figures/employee_age_distribution.png',
    bbox_inches='tight'
)

plt.show()

# Insight
print(f"Insight: The median employee age is {clean_df['Age'].median():.0f} years.")
```

```python

# 3. Scatter Plot – Monthly Income vs Total Working Years

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=clean_df,
    x='TotalWorkingYears',
    y='MonthlyIncome',
    hue='Attrition',
    palette={'No': '#4472C4', 'Yes': '#ED7D31'},
    alpha=0.7
)

plt.title('Monthly Income vs Total Working Years')
plt.xlabel('Total Working Years (Years)')
plt.ylabel('Monthly Income (Dataset Units)')
plt.legend(title='Attrition')

plt.savefig(
    '../Reports/figures/income_vs_working_years.png',
    bbox_inches='tight'
)

plt.show()

# Insight
correlation = clean_df['TotalWorkingYears'].corr(clean_df['MonthlyIncome'])
print(f"Insight: Monthly income and total working years have a correlation of {correlation:.2f}.")
```

```python

# 4. Line Chart – Average Monthly Income by Years at Company

income_by_experience = clean_df.groupby(
    'YearsAtCompany'
)['MonthlyIncome'].mean().reset_index()

income_by_experience = income_by_experience.sort_values(
    'YearsAtCompany'
)

plt.figure(figsize=(9, 5))

sns.lineplot(
    data=income_by_experience,
    x='YearsAtCompany',
    y='MonthlyIncome',
    color='#4472C4',
    marker='o'
)

plt.title('Average Monthly Income by Years at Company')
plt.xlabel('Years at Company (Years)')
plt.ylabel('Average Monthly Income (Dataset Units)')

plt.savefig(
    '../Reports/figures/income_by_years_at_company.png',
    bbox_inches='tight'
)

plt.show()

# Insight
highest_income_row = income_by_experience.loc[
    income_by_experience['MonthlyIncome'].idxmax()
]

print(
    f"Insight: The highest average monthly income occurs at "
    f"{int(highest_income_row['YearsAtCompany'])} years at the company."
)
```

```python

# 5. Box Plot – Monthly Income by Attrition

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=clean_df,
    x='Attrition',
    y='MonthlyIncome',
    hue='Attrition',
    palette={'No': '#4472C4', 'Yes': '#ED7D31'},
    legend=False
)

plt.title('Monthly Income Distribution by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Monthly Income (Dataset Units)')

plt.savefig(
    '../Reports/figures/monthly_income_by_attrition.png',
    bbox_inches='tight'
)

plt.show()

# Insight
median_income = clean_df.groupby('Attrition')['MonthlyIncome'].median()

print(
    f"Insight: Median monthly income was {median_income['No']:.0f} "
    f"for employees who stayed and {median_income['Yes']:.0f} "
    f"for employees who left."
)
```
